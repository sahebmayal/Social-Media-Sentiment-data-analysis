```python
# Pandas helps us work with tables and datasets.
#we will use it to read, clean and analyze our data.
import pandas as pd

#Matplotlib helps us create graphs and charts.
import matplotlib.pyplot as plt

#Seaborn helps us create better-looking statistical graphs.
import seaborn as sns
```

```python
#We are reading our CSV file and storing it inside a variable called "df".
# df meanes Dataframe - a table containing our data.

df = pd.read_csv("Social Media.csv")

# Let's look at the first 5 rows of our dataset.
df.head(2)
```

#Understand our Data

```python
# This tells us how many rows and columns our dataset has.

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Let's see the names of all our columns.

print(df.columns)
```

```python
# Let's see the type of information stored in each column.

df.info()
```

#Data Cleaning

```python
# These columns are old index numbers created by the original dataset.
# They don't give us useful information for our analysis.
# So we can remove them.

df = df.drop(columns=["Unnamed: 0.1", "Unnamed: 0"])

# Let's check our columns again.

df.columns
```

#Remove Extra Spaces

```python
# We will remove unnecessary spaces from our text-based columns.

df["Text"] = df["Text"].str.strip()
df["Sentiment"] = df["Sentiment"].str.strip()
df["User"] = df["User"].str.strip()
df["Platform"] = df["Platform"].str.strip()
df["Hashtags"] = df["Hashtags"].str.strip()

# Let's check the Sentiment column after cleaning.

df["Sentiment"].head()
```

#Convert Timestamp into Date and Time

```python
# Timestamp currently contains date and time information.
# We convert it into a proper datetime format so Python can understand it.

df["Timestamp"] = pd.to_datetime(df["Timestamp"], errors="coerce")

# Let's look at the result.

df["Timestamp"].head()
```

```python
# Extract the hour from the timestamp.
# For example, 18 means the post was made at 6 PM.

df["Hour"] = df["Timestamp"].dt.hour

# Extract the day of the week.

df["Day"] = df["Timestamp"].dt.day_name()

# Extract the month.

df["Month"] = df["Timestamp"].dt.month_name()

# Display the new columns.

df[["Timestamp", "Hour", "Day", "Month"]].head()
```

#Check for Missing Values

```python
# Missing values are empty pieces of information.
# We need to check whether our dataset contains any.

df.isnull().sum()
```

#Check for Duplicate Posts

```python
# Duplicate rows mean the same record may have been entered more than once.

print("Number of duplicate rows:", df.duplicated().sum())
```

If there are duplicates

```python
# Remove duplicate rows.

df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
```

# Basic Statistics

```python
# Describe gives us basic statistical information
# about our numerical columns such as Likes and Retweets.

df[["Likes", "Retweets"]].describe()
```


Our first question - What sentiments are present?

```python
# Let's see the different sentiment categories in our dataset.

print(df["Sentiment"].unique())
```

```python
# Count how many posts belong to each sentiment.

sentiment_counts = df["Sentiment"].value_counts(5)

print(sentiment_counts)
```

Graph: Sentiment Distribution

```python
# Create a bar chart showing the number of posts for each sentiment.

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Sentiment")

plt.title("Distribution of Social Media Sentiments")
plt.xlabel("Sentiment")
plt.ylabel("Number of Posts")

plt.xticks(rotation=45)

plt.show()
```

Which sentiment gets the most likes?

```python
# We calculate the average number of likes for each sentiment.

average_likes = df.groupby("Sentiment")["Likes"].mean()

print(average_likes)
```

```python
plt.figure(figsize=(8, 5))

average_likes.plot(kind="line")

plt.title("Average Likes by Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Average Likes")

plt.xticks(rotation=45)

plt.show()
```

Which sentiment gets the most retweets?

```python
# Calculate the average retweets for each sentiment.

average_retweets = df.groupby("Sentiment")["Retweets"].mean()

print(average_retweets)
```

```python
plt.figure(figsize=(8, 5))

average_retweets.plot(kind="line")

plt.title("Average Retweets by Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Average Retweets")

plt.xticks(rotation=45)

plt.show()
```

Which platform has the most posts?

```python
# Count the number of posts on each social-media platform.

platform_counts = df["Platform"].value_counts()

print(platform_counts)
```

```python
plt.figure(figsize=(8, 5))

platform_counts.plot(kind="pie")

plt.title("Number of Posts by Platform")
plt.xlabel("Platform")
plt.ylabel("Number of Posts")

plt.xticks(rotation=45)

plt.show()
```
